```python
text_columns = [
    "Product Category",
    "SubCategory",
    "Product",
    "Customer Gender",
    "Location",
    "Zone",
    "Delivery Type",
    "Status",
    "Reason"
]

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df["Delivery Days"] = (
    df["Delivery Date"] - df["OrderDate"]
).dt.days
```
